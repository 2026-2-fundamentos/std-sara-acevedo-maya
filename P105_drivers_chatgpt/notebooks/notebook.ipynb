# Horas y millas por conductor
Se suman las horas y las millas de cada conductor y se grafican los diez con mayor recorrido.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

activity = Path.cwd()
if activity.name == "notebooks":
    activity = activity.parent

drivers = pd.read_csv(activity / "data/drivers.csv")
timesheet = pd.read_csv(activity / "data/timesheet.csv")
totals = timesheet.groupby("driverId", as_index=False)[["hours-logged", "miles-logged"]].sum()
summary = totals.merge(drivers[["driverId", "name"]], on="driverId")

submission = activity / "submission"
submission.mkdir(exist_ok=True)
summary.to_csv(submission / "summary.csv", index=False)

top10 = summary.nlargest(10, "miles-logged").sort_values("miles-logged")
ax = top10.plot.barh(x="name", y="miles-logged", legend=False, xlabel="Millas", ylabel="Conductor")
ax.figure.savefig(submission / "top10_drivers.png", bbox_inches="tight")
plt.show()
```
